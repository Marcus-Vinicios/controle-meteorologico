```python
#https://docs.google.com/spreadsheets/d/1hRJ__svaFkactCc5rF1WGqmnmm6-lG2_lYWXv69v06w/edit?usp=sharing

import pandas as pd
import requests as rs
import time

endpoint_planilha = "https://docs.google.com/spreadsheets/d/1hRJ__svaFkactCc5rF1WGqmnmm6-lG2_lYWXv69v06w/export?format=csv"
endpoint_api = "https://api.open-meteo.com/v1/forecast"
```

```python
print('---- Dados Coletados ----')
data = pd.read_csv(endpoint_planilha)

for i, line in data.iterrows():
    name = line['Nome do local']
    lat = line['Latitude']
    lon = line['Longitude']

    params = {
        'latitude': lat,
        'longitude': lon,
        'current': 'temperature_2m'
    }

    response = rs.get(endpoint_api, params=params).json()

    display(response)
    print('\n\n')
```
